
# Q10. Remove Duplicate Records

## Problem Statement
Create a DataFrame:

```python
Name = ["Aman","Rahul","Aman","Neha"]
```

Check duplicate rows using `duplicated()` and remove them using `drop_duplicates()`.

## Sample Output
```python
Aman
Rahul
Neha
```


````python
import pandas as pd

# Create DataFrame
df = pd.DataFrame({
    "Name": ["Aman", "Rahul", "Aman", "Neha"]
})

# Remove duplicate names
df = df.drop_duplicates()

print(df)
````
